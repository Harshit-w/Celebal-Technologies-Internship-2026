```python
from pyspark.sql import SparkSession   # IMPORT LINRARIES
from pyspark.sql.functions import *
import pandas as pd
spark = SparkSession.builder \
    .appName("Week6Assignment") \
    .getOrCreate()
```

```python
excel_df = pd.read_csv("/content/drive/MyDrive/Dataset for week 6.csv")
excel_df.head() # READ DATSET
```

```python
df = spark.createDataFrame(excel_df)
df.show(10) # CREATE DATAFRAME
```

```python
df.printSchema() # VIEW SCHEMA AND VERIFY NUMBER OF RECORDS
print(df.count())
```

ASSIGNMENT  TASK FOR WEEK 6

```python
# Transformation
filtered_df = df.filter(df.amount > 1000)

# Transformation
selected_df = filtered_df.select("product_id", "amount")

# In spark's lazy evalution transformation do not work unless any action is defined.
selected_df.show(5)
```

```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .csv("/content/drive/MyDrive/Dataset for week 6.csv") # read the datset using header = true and inferschema = true
df.show(5)
```

```python
df.write.mode("overwrite").parquet("/content/drive/MyDrive/parquet_dataset") # create a parquet file
parquet_df = spark.read.parquet("/content/drive/MyDrive/parquet_dataset")
parquet_df.show(5) # Display data in parquet file which is same as csv file
```

```python
electronics_category_df = df.filter(col("category") == "Electronics") \
                   .select("product_id", "price")
electronics_category_df.show(7) # select columns product_id and price where the category is 'Electronics'
```

```python
new_df = df.withColumnRenamed("old_name", "new_name") \
               .withColumn("price", col("price").cast("double"))
new_df.show(5)   # renaming the column old_name to new_name and casting the price column from a String to a Double
```

```python
completed_status_df = df.filter((col("status") == "Completed") & (col("amount") > 1000))
completed_status_df.show(5)  # Select rows where the status is 'Completed' AND the amount is greater than 1000
```

```python
final_price_df = df.withColumn("final_price",col("base_price") * 1.18)
final_price_df.show(5)   # Create new column final_price which is the base_price multiplied by 1.18 (18% tax)
```

```python
df = spark.read.parquet("/content/drive/MyDrive/parquet_dataset")
filtered_df = df.filter(col("user_id").isNotNull())
filtered_df.write \
    .mode("overwrite") \
    .option("header", "true") \
    .csv("/content/drive/MyDrive/output_csv") # select Parquet file from "path/to/input", filter out any rows where user_id is null, and save the result as a CSV
```

```python
filter_df = df.filter((col("region") == "North") |(col("priority") == "High"))
filter_df.show(7) # Display rows where the region is 'North' OR the priority is 'High'
```
