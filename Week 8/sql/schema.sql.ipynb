```python
%sql
-- Schema Definition
CREATE TABLE IF NOT EXISTS customers (customer_id INT,customer_name STRING,email STRING,registration_date DATE,customer_type STRING)
USING DELTA;

CREATE TABLE IF NOT EXISTS products (product_id INT,product_name STRING,category STRING,subcategory STRING,cost_price DECIMAL(12,2))
USING DELTA;

CREATE TABLE IF NOT EXISTS orders (order_id INT,customer_id INT,order_date TIMESTAMP,status STRING,region_code INT)
USING DELTA;

CREATE TABLE IF NOT EXISTS order_items (item_id INT,order_id INT, product_id INT,quantity INT,unit_price DECIMAL(12,2),discount_percent DECIMAL(5,2))
USING DELTA;
```
